```python
import torch
import timm
from torch import nn
```

```python
def crear_modelo(arquitectura, num_classes, pretrained = True):                                                  # Poder pasar cualquier modelo soportado por timm
    dispositivo = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    modelo = timm.create_model(arquitectura, pretrained = pretrained, num_classes = num_classes)
    modelo.to(dispositivo)                                                                                            # Mover el modelo a la GPU
    return modelo

modelo_vit = crear_modelo("vit_base_patch16_224", num_classes = 37)

modelo_resnet = crear_modelo("resnet50", num_classes = 37)

loss_fn = nn.CrossEntropyLoss()                                                                                  # Calcula cuánto se equivoca el modelo al predecir vs la etiqueta real
optimizer = torch.optim.AdamW(modelo_vit.parameters(), lr=1e-4)                                                  # Optimizar los parámetros del modelo_vit. Tasa de aprendizaje: 0.0001

# Mostrar arquitectura relevante
print(modelo_vit.patch_embed)
print(modelo_vit.head)
```
